```python
# build a generic psim for recon2.2 machinery + ME machinery from gold-standard PSIM
# see "build_psim" notebook for how the gold-standard PSIM was constructed

# this will be used as input to the average_protein_features function in utils.functions
# when the context_specific parameter is set to False

import pandas as pd
import cobra

import sys
sys.path.insert(1, '../../')
from human_me.preprocess import correct_inputs as ci
```

```python
build_files_path = '/data2/hratch/human_me/build/'
psim_gold = pd.read_hdf(build_files_path + 'psim_me.h5')

input_data_path = '/data2/hratch/human_me/inputs/'
model = cobra.io.read_sbml_model(input_data_path + 'recon2_2.xml')
cm_1, cm_2, cm_3 = ci.correct_model(model)
```

```python
# filter for metabolic machinery only
metabolic_machinery = [g.id for g in cm_3.genes]
psim_gold = psim_gold[psim_gold.HGNC_ID.isin(metabolic_machinery)]
psim_gold.reset_index(inplace = True, drop = True)
psim_gold.to_csv(build_files_path + 'recon2_2_only_psim.csv')
```
